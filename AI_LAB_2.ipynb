```python
import random

class VacuumEnvironment:
    """Two-square world: locations 'A' and 'B', each Clean (0) or Dirty (1)."""

    def __init__(self):
        self.status = {'A': random.randint(0, 1), 'B': random.randint(0, 1)}
        self.agent_location = random.choice(['A', 'B'])

    def percept(self):
        return (self.agent_location, self.status[self.agent_location])

    def execute(self, action):
        if action == 'Suck':
            self.status[self.agent_location] = 0
        elif action == 'Right':
            self.agent_location = 'B'
        elif action == 'Left':
            self.agent_location = 'A'

def reflex_vacuum_agent(percept):
    """Condition-action rules: decide using ONLY the current percept."""
    location, dirty = percept
    if dirty:
        return 'Suck'
    elif location == 'A':
        return 'Right'
    else:
        return 'Left'

# Simulation setup
random.seed(7)
env = VacuumEnvironment()
env.status = {'A': 1, 'B': 1}
env.agent_location = 'B'

print("Initial state :", env.status, "| Agent at", env.agent_location)

score = 0
for step in range(1, 5):
    p = env.percept()
    action = reflex_vacuum_agent(p)
    env.execute(action)

    if action == 'Suck':
        score += 10       # reward for cleaning
    else:
        score -= 1        # cost of moving

    print(f"Step {step}: percept={p} -> action={action:5s} | world={env.status}")

print("Final state    :", env.status)
print("Performance score:", score)
```


Task 4: Model-based agent



```python
def model_based_agent(percept, memory):
    loc, dirty = percept
    if dirty:
        return 'Suck'
    memory.add(loc)                 # remember: this room is clean
    if memory == {'A', 'B'}:        # both rooms known clean
        return 'NoOp'               # stop, no penalty
    return 'Right' if loc == 'A' else 'Left'
```

```python
random.seed(42)
states = [make_state() for _ in range(10)]      # same 10 states for both agents

for name, agent in [('Reflex', reflex_agent), ('Model-based', model_based_agent)]:
    scores = [run(agent, dict(s), l) for s, l in states]
    print(name, 'average score =', st.mean(scores))
```

Post-lab: 3×3 grid

```python
import random
from collections import deque

moves_map = {'Up': (-1, 0), 'Down': (1, 0), 'Left': (0, -1), 'Right': (0, 1)}
```

```python
def next_move(pos, visited):
    queue = deque([(pos, [])])          # (position, path of moves so far)
    seen = {pos}
    while queue:
        p, path = queue.popleft()
        if p not in visited:            # found an unvisited room
            return path[0] if path else None
        for action, (dr, dc) in moves_map.items():
            n = (p[0] + dr, p[1] + dc)
            if 0 <= n[0] < 3 and 0 <= n[1] < 3 and n not in seen:
                seen.add(n)
                queue.append((n, path + [action]))
    return None                         # every room visited
```

```python
random.seed(42)                         # optional: repeatable results
dirt = {(r, c): random.randint(0, 1) for r in range(3) for c in range(3)}
pos = (random.randrange(3), random.randrange(3))

print("Initial dirt (1 = dirty):")
for r in range(3):
    print([dirt[(r, c)] for c in range(3)])
print("Agent starts at:", pos)
```

```python
visited = set()
moves = cleaned = score = 0

while True:
    if dirt[pos]:                       # dirty -> Suck
        dirt[pos] = 0
        cleaned += 1
        score += 10
        continue
    visited.add(pos)                    # remember this room is done
    action = next_move(pos, visited)
    if action is None:                  # all rooms visited -> NoOp
        break
    pos = (pos[0] + moves_map[action][0], pos[1] + moves_map[action][1])
    moves += 1
    score -= 1

print("Rooms visited :", len(visited), "/ 9")
print("Rooms cleaned :", cleaned)
print("Moves         :", moves)
print("Score         :", score)
print("Final dirt    :", [dirt[(r, c)] for r in range(3) for c in range(3)])
```
